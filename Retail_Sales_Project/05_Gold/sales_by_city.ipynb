```python
from pyspark.sql.functions import *

silver_df = spark.table(
    "retail_analytics_catalog.silver.sales_clean"
)

sales_by_city = (
    silver_df
    .groupBy(
        "city",
        "state"
    )
    .agg(
        sum("sales_amount")
        .alias("total_sales")
    )
)
```

```python
sales_by_city.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable(
"retail_analytics_catalog.gold.sales_by_city"
)
```
