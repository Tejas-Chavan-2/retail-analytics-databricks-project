Monthly Revenue

```python
from pyspark.sql.functions import *

silver_df = spark.table(
    "retail_analytics_catalog.silver.sales_clean"
)

monthly_sales = (
    silver_df
    .groupBy(
        "order_year",
        "order_month"
    )
    .agg(
        sum("sales_amount")
        .alias("monthly_revenue")
    )
)
```

```python
monthly_sales.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable(
"retail_analytics_catalog.gold.monthly_sales"
)
```
