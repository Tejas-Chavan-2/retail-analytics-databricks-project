Read Silver Data

```python
from pyspark.sql.functions import *

silver_df = spark.table(
    "retail_analytics_catalog.silver.sales_clean"
)
```

Aggregate

```python
sales_by_product = (
    silver_df
    .groupBy(
        "product_name",
        "category"
    )
    .agg(
        sum("sales_amount")
        .alias("total_sales"),
        
        sum("quantity")
        .alias("total_quantity")
    )
)
```

Save Gold Table

```python
sales_by_product.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable(
"retail_analytics_catalog.gold.sales_by_product"
)
```
