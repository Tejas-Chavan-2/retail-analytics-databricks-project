```python
sales_df = spark.read.csv(
    "/Volumes/retail_analytics_catalog/bronze/sales_volume/sales.txt",
    header=True,
    inferSchema=True
)

sales_df.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable(
    "retail_analytics_catalog.bronze.sales_raw"
)

print("sales_raw loaded successfully")
```
