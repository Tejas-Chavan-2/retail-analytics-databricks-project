```python
customer_df = spark.read.csv(
    "/Volumes/retail_analytics_catalog/bronze/sales_volume/customers.txt",
    header=True,
    inferSchema=True
)

customer_df.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable(
    "retail_analytics_catalog.bronze.customers_raw"
)

print("customers_raw loaded successfully")
```
