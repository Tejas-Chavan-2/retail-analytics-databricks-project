```python
product_df = spark.read.csv(
    "/Volumes/retail_analytics_catalog/bronze/sales_volume/products.txt",
    header=True,
    inferSchema=True
)

product_df.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable(
    "retail_analytics_catalog.bronze.products_raw"
)

print("products_raw loaded successfully")
```
