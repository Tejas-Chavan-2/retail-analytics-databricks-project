```python
from pyspark.sql import Row
from datetime import datetime

def write_log(layer,status,message):

    log_df = spark.createDataFrame([
        Row(
            log_time=datetime.now(),
            pipeline_name="Retail_Sales_Pipeline",
            layer=layer,
            status=status,
            message=message
        )
    ])

    log_df.write.mode("append").saveAsTable(
        "retail_analytics_catalog.silver.pipeline_logs"
    )
```
