# Medallion Architecture

## Bronze Layer

Purpose:

Store raw source data exactly as received.

Tables:

- customers_raw
- products_raw
- sales_raw

Activities:

- File ingestion
- Schema inference
- Delta table creation

No transformations are performed in the Bronze layer.

---

## Validation Layer

Purpose:

Validate source data before processing.

Validation Checks:

- Null Validation
- Duplicate Validation
- Negative Quantity Validation
- Invalid Customer Validation
- Invalid Product Validation

Invalid records are stored in:

silver.rejected_records

---

## Silver Layer

Purpose:

Store clean and enriched data.

Transformations:

- Remove invalid records
- Remove duplicate records
- Customer master join
- Product master join
- sales_amount calculation
- order_month calculation
- order_year calculation

Output Table:

sales_clean

---

## Gold Layer

Purpose:

Provide business-ready datasets for reporting and analytics.

Tables:

### sales_by_product

Product sales performance analysis.

### sales_by_city

Regional sales performance analysis.

### top_customers

Customer ranking based on revenue.

### monthly_sales

Monthly revenue trend analysis.

# Interview Summary

I built an end-to-end Retail Sales Analytics Platform using Azure Databricks and Unity Catalog.

Data was ingested from CSV files stored in Unity Catalog Volumes into Bronze Delta tables.

A validation framework was implemented to perform:

- Null Validation
- Duplicate Validation
- Invalid Customer Validation
- Invalid Product Validation
- Negative Quantity Validation

Invalid records were stored separately in a rejected records table.

Valid records were transformed and enriched in the Silver layer using customer and product master data.

Gold layer tables were created for business reporting including:

- Sales by Product
- Sales by City
- Top Customers
- Monthly Revenue

The complete pipeline was orchestrated using Databricks Workflows with retry logic, logging, and Delta Lake optimization while leveraging Unity Catalog for governance and centralized management.

# Production Support Approach

When a failure occurs:

1. Workflow detects failure.
2. Retry mechanism starts.
3. Failure logged in pipeline_logs.
4. Email notification sent.
5. Support team investigates root cause.
6. Pipeline rerun after fix.

Benefits:

- Improved Reliability
- Faster Recovery
- Reduced Manual Intervention
- Better Monitoring
