```python
%sql
CREATE CATALOG IF NOT EXISTS retail_analytics_catalog;

CREATE SCHEMA IF NOT EXISTS retail_analytics_catalog.bronze;

CREATE SCHEMA IF NOT EXISTS retail_analytics_catalog.silver;

CREATE SCHEMA IF NOT EXISTS retail_analytics_catalog.gold;
```

```python
%sql
CREATE VOLUME IF NOT EXISTS retail_analytics_catalog.bronze.sales_volume;
```

```python
%sql
CREATE TABLE IF NOT EXISTS retail_analytics_catalog.silver.pipeline_logs
(
    log_time TIMESTAMP,
    pipeline_name STRING,
    layer STRING,
    status STRING,
    message STRING
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS retail_analytics_catalog.silver.audit_table
(
    run_id STRING,
    table_name STRING,
    source_count BIGINT,
    target_count BIGINT,
    status STRING,
    load_time TIMESTAMP
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS retail_analytics_catalog.silver.rejected_records
(
    order_id STRING,
    customer_id STRING,
    product_id STRING,
    quantity STRING,
    order_date STRING,
    reason STRING,
    load_time TIMESTAMP
);
```
